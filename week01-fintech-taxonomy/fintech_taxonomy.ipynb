## 1. Dataset & Classification

First, we will load a mock FinTech companies dataset representing global companies (simulating a Kaggle FinTech dataset export) into a pandas DataFrame. We will then manually classify 25 of these companies into the following 7-category taxonomy:
* Payments & Digital Wallets
* Digital Banking & Neobanks
* Lending Tech & BNPL
* WealthTech & Robo-Advisory
* InsurTech
* Blockchain & Crypto
* RegTech

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Create a representative FinTech dataset of 25 companies
data = {
    'Company': [
        'Stripe', 'Revolut', 'Klarna', 'Robinhood', 'Lemonade',
        'Coinbase', 'Chainalysis', 'Adyen', 'Monzo', 'Affirm',
        'Betterment', 'Oscar Health', 'Binance', 'IdentityMind', 'Wise',
        'Chime', 'SoFi', 'Wealthfront', 'Root Insurance', 'Ledger',
        'ComplyAdvantage', 'PayPal', 'N26', 'Afterpay', 'Carta'
    ],
    'Description': [
        'Payment processing infrastructure for the internet', 'Digital banking app with currency exchange', 'Buy now pay later shopping service', 'Commission-free stock and crypto investing', 'AI-driven digital insurance platform',
        'Cryptocurrency exchange platform', 'Blockchain data platform and compliance', 'Global payment platform for merchants', 'UK digital neobank', 'Lending platform for point-of-sale checkout',
        'Robo-advisor for automated investing', 'Tech-driven health insurance plans', 'Global crypto exchange ecosystem', 'SaaS platform for AML and risk mitigation', 'International money transfer platform',
        'US digital banking platform', 'Personal finance platform with student loans and investing', 'Automated investment service', 'Telematics-based car insurance', 'Hardware wallets for cryptocurrency storage',
        'AI-driven financial crime detection and AML compliance', 'Digital wallet and payment processor', 'European digital neobank', 'Installment payment provider', 'Equity management and valuation platform'
    ]
}

df = pd.DataFrame(data)

# Map each company to the 7-category taxonomy
taxonomy_mapping = {
    'Stripe': 'Payments & Digital Wallets',
    'Revolut': 'Digital Banking & Neobanks',
    'Klarna': 'Lending Tech & BNPL',
    'Robinhood': 'WealthTech & Robo-Advisory',
    'Lemonade': 'InsurTech',
    'Coinbase': 'Blockchain & Crypto',
    'Chainalysis': 'RegTech',
    'Adyen': 'Payments & Digital Wallets',
    'Monzo': 'Digital Banking & Neobanks',
    'Affirm': 'Lending Tech & BNPL',
    'Betterment': 'WealthTech & Robo-Advisory',
    'Oscar Health': 'InsurTech',
    'Binance': 'Blockchain & Crypto',
    'IdentityMind': 'RegTech',
    'Wise': 'Payments & Digital Wallets',
    'Chime': 'Digital Banking & Neobanks',
    'SoFi': 'Digital Banking & Neobanks',
    'Wealthfront': 'WealthTech & Robo-Advisory',
    'Root Insurance': 'InsurTech',
    'Ledger': 'Blockchain & Crypto',
    'ComplyAdvantage': 'RegTech',
    'PayPal': 'Payments & Digital Wallets',
    'N26': 'Digital Banking & Neobanks',
    'Afterpay': 'Lending Tech & BNPL',
    'Carta': 'RegTech'
}

df['Taxonomy Category'] = df['Company'].map(taxonomy_mapping)
display(df)
```

### Classification Justification (3 borderline cases)

1. **SoFi**: While SoFi started primarily as a peer-to-peer student loan refinancing company (which fits *Lending Tech & BNPL*), it was classified as **Digital Banking & Neobanks**. This is because it obtained a full banking charter and currently offers standard checking/savings deposit accounts, debit cards, and an integrated banking application, making its core model modernly aligned with a neobank.
2. **Wise**: Initially known as TransferWise, its core feature is cross-border money transfer. This could overlap with *Digital Banking* because of its multi-currency accounts. However, it was classified under **Payments & Digital Wallets** because its primary economic utility is payment routing/remittance rather than acting as a full lending or deposit institution.
3. **Carta**: Carta provides equity management, cap table software, and private market valuations. While it touches financial reporting, it is classified as **RegTech** because its value proposition relies heavily on helping companies comply with complex securities regulations, financial reporting standards, and equity tax codes.

## 3. Visualization

We will visualize the distribution of our classified global fintech companies across the 7 categories.

```python
category_counts = df['Taxonomy Category'].value_counts()

plt.figure(figsize=(10, 6))
colors = ['#1f77b4', '#aec7e8', '#ff7f0e', '#ffbb78', '#2ca02c', '#98df8a', '#d62728']
category_counts.plot(kind='bar', color=colors, edgecolor='black')

plt.title('FinTech Companies Count by Taxonomy Category', fontsize=14, fontweight='bold')
plt.xlabel('Taxonomy Category', fontsize=12)
plt.ylabel('Number of Companies', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Interpretation
In the global dataset, **Digital Banking & Neobanks** and **Payments & Digital Wallets** are the most represented categories. This heavily matches what we expect from the Pakistani market discussed in Week 1, where the initial financial infrastructure layer has historically been dominated by mobile wallets (Easypaisa, JazzCash) and EMIs (SadaPay, NayaPay) striving to bridge the massive unbanked population gap.
